Problem 1: Unique Words in a Sentence

```python
# Taking input sentence
sentence = "python is easy and python is powerful"

# Splitting sentence into words
words = sentence.split()

# Converting list to set to get unique words
unique_words = set(words)

# Printing reesults
print("Unique words count:", len(unique_words))
print("Unique words:", unique_words)
```

Problem 2: Highest Salary from Employee Data

```python
# Employee salary data
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

# Assuming first employee has the highest salary initially
highest_employee = ""
highest_salary = 0

# Checking each employee's salary
for name, salary in employees.items():
    if salary > highest_salary:
        highest_salary = salary
        highest_employee = name

print("Highest Salary:", highest_employee, "-", highest_salary)        
```

Problem 3: Find maximum and minimum values from a list of numbers.

```python
# Given list of nos.
numbers = [45, 22, 89, 10, 66]

# Initialising max and min with the first element
maximum = numbers[0]
minimum = numbers[0]

# Loop through the list 
for num in numbers:
    if num > maximum:
        maximum = num
    if num < minimum:
        minimum = num
    
#Printing results
print("List:",numbers)
print("Max:",maximum)
print("Min:",minimum)
```

Problem: Count Products with Price Greater Than 1000

```python
# List of product prices
prices = [450, 1200, 899, 1500, 300]

count = 0

# Checking each price
for price in prices:
    if price > 1000:
        count += 1

# Printing the result
print("Products above 1000:", count)
```

Problem 5: Calculate Attendance Percentage

```python
# Attendance list
attendance = ["P", "P", "A", "P", "P"]

present_count = 0
total_days = len(attendance)

# Counting number of present days
for day in attendance:
    if day == "P":
        present_count += 1

# Calculating attendance percentage
attendance_percentage = (present_count / total_days) * 100

# Printing result
print("Attendance Percentage:", attendance_percentage)
```

Problem 7: Count Character Frequency

```python
# Given string
text = "python"

char_count = {}

# Counting each character
for ch in text:
    if ch in char_count:
        char_count[ch] += 1
    else:
        char_count[ch] = 1

# Printing the result
print(char_count)

```

Problem 9: Check if a Key Exists in a Dictionary

```python
# Given dictionary
employee = {
    "name": "Ravi",
    "age": 25,
    "department": "IT"
}

# Key to check
key_to_check = "name"

# Checking if key exists
if key_to_check in employee:
    print("Employee exists")
else:
    print("Employee does not exist")
```

Problem 10: Calculate Average Marks

```python
# List of marks
marks = [70, 75, 80, 65]

total = 0

# Calculating total marks
for m in marks:
    total += m

# Calculating average
average = total / len(marks)

# Printing result
print("Average Marks:", average)
```
